```python
import pandas as pd

df = pd.read_csv("/content/drive/MyDrive/Modulo6_DCDD/creditcard.csv")
df.head()
```

```python
import numpy as np

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import precision_score, recall_score, f1_score, roc_auc_score, accuracy_score

import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Dropout
from tensorflow.keras.callbacks import EarlyStopping

scaler = StandardScaler()

df["Time"] = scaler.fit_transform(df[["Time"]])
df["Amount"] = scaler.fit_transform(df[["Amount"]])

X = df.drop("Class", axis=1)
y = df["Class"]

X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

X_train, X_val, y_train, y_val = train_test_split(
    X_train, y_train,
    test_size=0.20,
    random_state=42,
    stratify=y_train
)

early_stop = EarlyStopping(
    monitor="val_loss",
    patience=5,
    restore_best_weights=True
)

print("Datos listos")
print("Entrenamiento:", X_train.shape)
print("Validación:", X_val.shape)
print("Prueba:", X_test.shape)
```

```python
def registrar_modelo(nombre_modelo, modelo, parametros):
    with mlflow.start_run(run_name=nombre_modelo):
        mlflow.log_params(parametros)

        modelo.compile(
            optimizer="adam",
            loss="binary_crossentropy",
            metrics=["accuracy"]
        )

        history = modelo.fit(
            X_train,
            y_train,
            validation_data=(X_val, y_val),
            epochs=30,
            batch_size=32,
            callbacks=[early_stop],
            verbose=1
        )

        y_pred = (modelo.predict(X_test) > 0.5).astype(int)

        acc = accuracy_score(y_test, y_pred)
        precision = precision_score(y_test, y_pred)
        recall = recall_score(y_test, y_pred)
        f1 = f1_score(y_test, y_pred)
        roc_auc = roc_auc_score(y_test, y_pred)

        mlflow.log_metric("test_accuracy", acc)
        mlflow.log_metric("test_precision", precision)
        mlflow.log_metric("test_recall", recall)
        mlflow.log_metric("test_f1_score", f1)
        mlflow.log_metric("test_roc_auc", roc_auc)

        for epoch in range(len(history.history["loss"])):
            mlflow.log_metric("loss", history.history["loss"][epoch], step=epoch)
            mlflow.log_metric("accuracy", history.history["accuracy"][epoch], step=epoch)
            mlflow.log_metric("val_loss", history.history["val_loss"][epoch], step=epoch)
            mlflow.log_metric("val_accuracy", history.history["val_accuracy"][epoch], step=epoch)

        print(nombre_modelo)
        print("Accuracy:", acc)
        print("Precision:", precision)
        print("Recall:", recall)
        print("F1-score:", f1)
        print("ROC-AUC:", roc_auc)

        return history
```

## Modelo 1

```python
modelo_1 = Sequential([
    Dense(64, activation="relu", input_shape=(X_train.shape[1],)),
    Dense(1, activation="sigmoid")
])

history_1 = registrar_modelo(
    "Modelo_1_una_capa",
    modelo_1,
    {
        "capas_ocultas": 1,
        "neuronas": "64",
        "dropout": "No",
        "optimizer": "adam",
        "loss": "binary_crossentropy"
    }
)
```

## Modelo 2

```python
modelo_2 = Sequential([
    Dense(128, activation="relu", input_shape=(X_train.shape[1],)),
    Dense(64, activation="relu"),
    Dense(1, activation="sigmoid")
])

history_2 = registrar_modelo(
    "Modelo_2_dos_capas",
    modelo_2,
    {
        "capas_ocultas": 2,
        "neuronas": "128, 64",
        "dropout": "No",
        "optimizer": "adam",
        "loss": "binary_crossentropy"
    }
)
```

## Modelo 3

```python
modelo_3 = Sequential([
    Dense(128, activation="relu", input_shape=(X_train.shape[1],)),
    Dropout(0.3),
    Dense(64, activation="relu"),
    Dense(32, activation="relu"),
    Dense(1, activation="sigmoid")
])

history_3 = registrar_modelo(
    "Modelo_3_tres_capas_dropout",
    modelo_3,
    {
        "capas_ocultas": 3,
        "neuronas": "128, 64, 32",
        "dropout": "0.3",
        "optimizer": "adam",
        "loss": "binary_crossentropy"
    }
)
```
